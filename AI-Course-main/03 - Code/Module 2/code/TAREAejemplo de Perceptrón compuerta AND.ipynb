```python
# ============================================================
# TAREA
# 1. Investigar: Función de Activación (Entender su uso en Redes neuronales).
# 2. Investigar y Ejecutar un ejemplo de Perceptrón (tratar de entenderlo).
# ============================================================
#
# ---------- CONCEPTOS A TENER A MANO ----------
#
# 1. FUNCIÓN DE ACTIVACIÓN
#    - Se aplica a la salida de cada neurona: salida = f(suma_ponderada).
#    - Decide si la neurona "se activa" y cuánto.
#    - Su función principal es introducir NO LINEALIDAD: sin ella, aunque
#      se apilen muchas capas, la red sería equivalente a una sola
#      operación lineal y no aprendería patrones complejos.
#    - Tipos comunes:
#        * Escalón : da 0 o 1 (la del perceptrón clásico).
#        * Sigmoide: comprime a (0, 1), útil para probabilidades.
#        * Tanh    : comprime a (-1, 1).
#        * ReLU    : max(0, z), la más usada en capas ocultas.
#        * Softmax : convierte salidas en probabilidades que suman 1.
#
# 2. PERCEPTRÓN
#    - Es la neurona artificial más simple.
#    - Pasos: multiplica entradas por pesos, suma un bias y aplica
#      la función de activación.
#    - Pesos: la importancia de cada entrada.
#    - Bias  : ajusta el umbral a partir del cual se activa.
#    - Aprende ajustando pesos y bias cuando se equivoca:
#         w = w + tasa_aprendizaje * error * entrada
#    - Límite: solo resuelve problemas linealmente separables
#      (AND y OR sí; XOR no, para eso se necesitan capas ocultas).
# ============================================================

import numpy as np

# --- DATOS DE ENTRENAMIENTO (Compuerta AND) ---
# Cada fila de X es una entrada (entrada1, entrada2)
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
# Salida correcta para cada fila (AND: solo (1,1) da 1)
y = np.array([0, 0, 0, 1])

# --- PARÁMETROS INICIALES ---
w = np.zeros(2)   # pesos: empiezan en 0 (la neurona no sabe nada)
b = 0.0           # bias: empieza en 0
lr = 0.1          # tasa de aprendizaje: qué tanto corrige en cada error


# --- FUNCIÓN DE ACTIVACIÓN (Tipo Escalón) ---
def escalon(z):
    # Si la suma es mayor a 0, la neurona se activa (1). Si no, se apaga (0).
    # Se usa > y no >= para evitar errores de redondeo justo en el borde.
    return 1 if z > 0 else 0


# --- ENTRENAMIENTO ---
# Una "época" es una pasada completa por todos los datos
for epoca in range(20):
    errores = 0

    for xi, yi in zip(X, y):
        # 1. Suma ponderada: entradas * pesos + bias
        suma = np.dot(w, xi) + b

        # 2. Pasamos la suma por la función de activación
        pred = escalon(suma)

        # 3. Error: respuesta correcta menos lo que predijo
        #    (0 si acertó, 1 o -1 si falló)
        error = yi - pred

        # 4. Ajuste: si hubo error, mueve pesos y bias hacia la respuesta
        #    correcta. Si no hubo error, no cambia nada.
        w += lr * error * xi
        b += lr * error

        errores += abs(error)

    # Mostramos cómo van los pesos en cada época
    print(f"Época {epoca + 1}: w={w}, b={b:.1f}, errores={errores}")

    # Si ya no hubo errores, el perceptrón aprendió y paramos
    if errores == 0:
        break


# --- PRUEBA FINAL ---
print("\nProbando el Perceptrón (Compuerta AND):")
for xi in X:
    print("Entrada:", xi, "-> Salida:", escalon(np.dot(w, xi) + b))
```
